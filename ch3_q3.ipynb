Model: Salary = β0 + β1·GPA + β2·IQ + β3·Level + β4·(GPA×IQ) + β5·(GPA×Level), with Level = 1 for College, 0 for High School.

```python
import numpy as np
```

```python
b0, b1, b2, b3, b4, b5 = 50, 20, 0.07, 35, 0.01, -10  # fitted coefficients
```

**(a)** Compare high-school vs. college graduates at fixed GPA and IQ.

For College (Level=1): Salary = (β0+β3) + (β1+β5)·GPA + β2·IQ + β4·(GPA×IQ)

For High School (Level=0): Salary = β0 + β1·GPA + β2·IQ + β4·(GPA×IQ)

College earns more when (β3 + β5·GPA) > 0. Solve for the break-even GPA:

```python
breakeven_gpa = -b3 / b5  # GPA at which college and high-school salaries are equal
breakeven_gpa
```

Since β5 is negative, (β3 + β5·GPA) > 0 only when GPA is **below** this break-even point, and it turns negative once GPA exceeds it. So college graduates earn more only for low GPA; **high school graduates earn more once GPA is high enough** — answer **(iii)**.

**(b)** Predict salary for a college graduate with IQ = 110, GPA = 4.0.

```python
gpa, iq, level = 4.0, 110, 1  # college graduate
salary = (b0 + b1*gpa + b2*iq + b3*level
          + b4*gpa*iq + b5*gpa*level)  # in thousands of dollars
salary
```

**(c)** True or false: a small coefficient means little evidence of interaction?

**False.** Statistical significance of an interaction term depends on its standard error and resulting p-/t-value, not on the magnitude of the coefficient alone — GPA and IQ are on very different scales, so β4 = 0.01 can still be significant. This model's coefficients alone (without standard errors) can't confirm or rule out a real interaction effect.
