For K=1, the predicted class of any point is the class of its single nearest neighbor in the training set. When evaluating a training observation $x_i$, its nearest neighbor among the training set is $x_i$ itself, at distance 0.

Given training and test error for 1-NN is 18%:  

(Training error + Test error) / 2 = 18%     
(0% + Test error) / 2 = 18%     
**Test error (1-NN) = 36%**

### Comparing test errors

**Logistic regression:** 20% (Training Error Rate) | 30% (Test Error Rate)  
**1-NN:** 0% (Training Error Rate) | 36% (Test Error Rate)

Test error estimates performance on new, unseen observations    
training error is not a reliable guide to generalization, it can drive training error to 0 simply by memorizing, as 1-NN does by construction.

we should prefer logistic regression, since its test error (30%) is lower than 1-NN's test error (36%), even though 1-NN's has avg error rate of 18%, it is misleading because it was dragged down by 1-NN's 0% training error, which reflects memorization of the training set rather than any genuine predictive ability on new data.
