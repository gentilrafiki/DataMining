## 4.7.2 Logistic Regression

```python
allvars = Smarket.columns.drop(['Today', 'Direction', 'Year'])
design = MS(allvars)
X = design.fit_transform(Smarket)
y = Smarket.Direction == 'Up'
glm = sm.GLM(y,
             X,
             family=sm.families.Binomial())
results = glm.fit()
summarize(results)
```

```python
results.params
```

```python
results.pvalues
```

```python
probs = results.predict()
probs[:10]
```

```python
labels = np.array(['Down']*1250)
labels[probs > 0.5] = "Up"
```

```python
confusion_table(labels, Smarket.Direction)
```

### Train/Test split — fit on 2001-2004, test on 2005

So far we've trained and tested on the *same* 1,250 observations, which gives an overly optimistic error rate. Now we hold out 2005 as a genuine test set.

```python
train = (Smarket.Year < 2005)
Smarket_train = Smarket.loc[train]
Smarket_test = Smarket.loc[~train]
Smarket_test.shape
```

```python
X_train, X_test = X.loc[train], X.loc[~train]
y_train, y_test = y.loc[train], y.loc[~train]
glm_train = sm.GLM(y_train,
                    X_train,
                    family=sm.families.Binomial())
results = glm_train.fit()
probs = results.predict(exog=X_test)
```

```python
D = Smarket.Direction
L_train, L_test = D.loc[train], D.loc[~train]
```

```python
labels = np.array(['Down']*252)
labels[probs > 0.5] = 'Up'
confusion_table(labels, L_test)
```

```python
np.mean(labels == L_test), np.mean(labels != L_test)
```

### Refit using only Lag1 and Lag2

The full model had weak p-values across the board. Let's try a leaner model using just the two strongest predictors.

```python
model = MS(['Lag1', 'Lag2']).fit(Smarket)
X = model.transform(Smarket)
X_train, X_test = X.loc[train], X.loc[~train]
glm_train = sm.GLM(y_train,
                    X_train,
                    family=sm.families.Binomial())
results = glm_train.fit()
probs = results.predict(exog=X_test)
labels = np.array(['Down']*252)
labels[probs > 0.5] = 'Up'
confusion_table(labels, L_test)
```

### KNN on the Caravan data set

85 predictors, 5,822 individuals, predicting whether someone purchases caravan insurance. Only 6% of people purchased — a good example of why overall accuracy alone can be misleading on imbalanced data.

```python
Caravan = load_data('Caravan')
Purchase = Caravan.Purchase
Purchase.value_counts()
```

```python
348 / 5822
```

```python
feature_df = Caravan.drop(columns=['Purchase'])
```

KNN is distance-based, so the *scale* of each variable matters a lot. We standardize every predictor to mean 0, standard deviation 1 first.

```python
scaler = StandardScaler(with_mean=True,
                         with_std=True,
                         copy=True)
```

```python
scaler.fit(feature_df)
X_std = scaler.transform(feature_df)
```

```python
feature_std = pd.DataFrame(
                   X_std,
                   columns=feature_df.columns)
feature_std.std()
```

```python
(X_train,
 X_test,
 y_train,
 y_test) = train_test_split(feature_std,
                             Purchase,
                             test_size=1000,
                             random_state=0)
```

```python
knn1 = KNeighborsClassifier(n_neighbors=1)
knn1_pred = knn1.fit(X_train, y_train).predict(X_test)
np.mean(y_test != knn1_pred), np.mean(y_test != "No")
```

```python
confusion_table(knn1_pred, y_test)
```

```python
9/(53+9)
```

### Tuning K — trying several values with a loop

K is a *hyperparameter* — we don't know the best value in advance, so we try a range and compare.

```python
for K in range(1, 6):
    knn = KNeighborsClassifier(n_neighbors=K)
    knn_pred = knn.fit(X_train, y_train).predict(X_test)
    C = confusion_table(knn_pred, y_test)
    templ = ('K={0:d}: # predicted to purchase: {1:>2},' +
             ' # who did purchase {2:d}, accuracy {3:.1%}')
    pred = C.loc['Yes'].sum()
    did_purchase = C.loc['Yes', 'Yes']
    print(templ.format(
        K,
        pred,
        did_purchase,
        did_purchase / pred))
```

### Comparison to Logistic Regression (sklearn)

```python
logit = LogisticRegression(C=1e10, solver='liblinear')
logit.fit(X_train, y_train)
logit_pred = logit.predict_proba(X_test)
logit_labels = np.where(logit_pred[:, 1] > 0.5, 'Yes', 'No')
confusion_table(logit_labels, y_test)
```

With the default 0.5 cutoff, the model never predicts "Yes" at all. Lowering the threshold to 0.25 gives far more useful results:

```python
logit_labels = np.where(logit_pred[:, 1] > 0.25, 'Yes', 'No')
confusion_table(logit_labels, y_test)
```
